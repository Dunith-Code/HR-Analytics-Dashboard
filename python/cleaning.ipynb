```python
import pandas as pd
import os
```

```python
RAW_PATH = r"D:\Z\my-projects\data-science\HR-Analytics-Dashboard\data\raw\WA_Fn-UseC_-HR-Employee-Attrition.csv"
PROCESSED_PATH = r"D:\Z\my-projects\data-science\HR-Analytics-Dashboard\data\processed/hr_data_cleaned.csv"
```

```python
# LOAD
df = pd.read_csv(RAW_PATH)
print(f"Loaded {df.shape[0]} rows, {df.shape[1]} columns")
```

```python
# Drop constant
cols_to_drop = ["EmploymentCount", "Over18", "StandardHours"]
df = df.drop(columns=[c for c in cols_to_drop if c in df.columns])
print(f"Dropped columns: {cols_to_drop}")
```

```python
# Map numeric codes to labels
education_map = {1: "Below College", 2: "College", 3: "Bachelor", 4: "Master", 5: "Doctor"}
satisfaction_map = {1: "Bad", 2: "Medium", 3: "High", 4: "Very High"}
workfile_map = {1: "Bad", 2: "Good", 3: "Better", 4: "Best"}
performance_map = {1: "Low", 2: "Good", 3: "Excellent", 4: "Outstanding"}

df["EducationLabel"] = df["Education"].map(education_map)
df["EnvironmentSatisfactionLabel"] = df["EnvironmentSatisfaction"].map(satisfaction_map)
df["JobInvolvementLabel"] = df["JobInvolvement"].map(satisfaction_map)
df["JobSatisfactionLabel"] = df["JobSatisfaction"].map(satisfaction_map)
df["RelationshipSatisfactionLabel"] = df["RelationshipSatisfaction"].map(satisfaction_map)
df["WorkLifeBalanceLabel"] = df["WorkLifeBalance"].map(workfile_map)
df["PerformanceRatingLabel"] = df["PerformanceRating"].map(performance_map)
```

```python
# Derived columns for Power BI grouping
bins = [17, 25, 35, 45, 55, 65]
labels = ["18-25", "26-35", "36-45", "46-55", "56-65"]
df["AgeBand"] = pd.cut(df["Age"], bins=bins, labels=labels)

income_bins = [0, 3000, 6000, 10000, 15000, 20000]
income_labels = ["<3K", "3K-6K", "6K-10K", "10K-15K", "15K+"]
df["IncomeBand"] = pd.cut(df["MonthlyIncome"], bins=income_bins, labels=income_labels)
```

```python
# Save
os.makedirs(os.path.dirname(PROCESSED_PATH), exist_ok=True)
df.to_csv(PROCESSED_PATH, index=False)
print(f"Saved cleaned file to {PROCESSED_PATH}")
print(f"Final shape: {df.shape[0]} rows, {df.shape[1]} columns")
```
